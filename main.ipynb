```python
import numpy as np
from constants import DAY
from params import P, BANDS
from derived import derived, print_params
from lightcurves import lightcurves, check_energy, check_ssa
from thermal import thermal_and_check, thermal_band_luminosities
from plotting import make_plots

D = derived(P)
print_params(P, D)

print("\ncomputing synchrotron bands + SSA (%d time steps)..." % 400)
th, Lint, Lssa, TauSSA, Lnth = lightcurves(P, D)

print("\nsolving the thermal reservoir ODE...")
Eth, Lth, tau = thermal_and_check(th, Lnth, P, D)
Lth_band, Teff = thermal_band_luminosities(th, Lth, D["Rs0"] * th**P["n"], P)

check_energy(th, Lnth, P, D)
check_ssa(th, TauSSA, P)

figs, Lnth_out, Ltot = make_plots(th, P, D, Lssa, Lth_band, Teff, tau)

t = th * P["t_star"]
print("\nNon-thermal peaks:")
for b, (name, *_rest) in enumerate(BANDS):
    if Lnth_out[b].max() > 0:
        print("  %-26s peak L = %.2e erg/s at t = %.2f d" % (name, Lnth_out[b].max(), t[Lnth_out[b].argmax()] / DAY))
print("\nTotal (non-thermal+thermal) peaks:")
for b, (name, *_rest) in enumerate(BANDS):
    if Ltot[b].max() > 0:
        print("  %-26s peak L = %.2e erg/s at t = %.2f d" % (name, Ltot[b].max(), t[Ltot[b].argmax()] / DAY))

# selftest()  
```
